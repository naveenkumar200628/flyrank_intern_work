## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

### Task Type

This project is formulated as a **ranking problem using a classification model**.

The model predicts whether a page is at risk of decline, and the predicted probability scores are used to rank pages. This helps SEO analysts prioritize which pages should be reviewed and refreshed first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

### Target Variable

**Target:** At-Risk Page (1) vs Stable Page (0)

A page is labeled as **1 (At Risk)** if its CTR or traffic decreases by more than 25% when compared to the previous 30-day period.

Otherwise, it is labeled as **0 (Stable)**.

This target serves as a proxy for identifying pages that require analyst attention.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

### Why Machine Learning?

A rule-based approach such as:

CTR Drop > 25% → At Risk

is simple but fails to capture complex relationships between multiple SEO signals.

Machine Learning can learn patterns from:

- CTR trends
- Impression changes
- Ranking position movements
- Historical performance
- Seasonal behavior
- Interactions between multiple metrics

This allows the system to identify at-risk pages more accurately than a fixed threshold-based solution.



### Project Objective

Build a classification-based ranking system that identifies SEO pages at risk of decline and prioritizes them using probability scores.

The system helps SEO analysts focus on the most impactful pages first, reducing manual effort and improving decision-making efficiency.
